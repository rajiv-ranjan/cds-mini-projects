#### Preprocessing and cleaning the tweets

* Convert the text to lower case
* Remove usernames, hashtags and links from the text (tweets)

```python
# YOUR CODE HERE
# Define a cleaning function
def clean_tweet(text):
    if text:
        text = text.lower()  # Convert to lowercase
        text = re.sub(r'@\w+', '', text)  # Remove usernames
        text = re.sub(r'#\w+', '', text)  # Remove hashtags
        text = re.sub(r'http\S+|www\S+', '', text)  # Remove URLs
        text = re.sub(r'[^a-zA-Z0-9\s]', '', text)  # Remove special characters
        text = re.sub(r'\s+', ' ', text).strip()  # Remove extra spaces
        return text
    return ""
```

```python
# Register as UDF
clean_udf = udf(clean_tweet, StringType())
```

```python
# apply to DataFrame
df_cleaned = df_cleaned.withColumn("clean_text", clean_udf(col("text")))
```

### Making predictions using NaiveBayes

```python
def make_prediction_using_NaiveBayes(user_input):
    # Create a Spark DataFrame from the user input
    df1 = spark.createDataFrame([(1, user_input)], ['Id', 'text'])

    # Apply the same preprocessing steps as the training data
    df1 = df1.withColumn("clean_text", clean_udf(col("text")))
    df1 = df1.withColumn("tokens", tokenize_udf(col("clean_text")))
    df1 = df1.withColumn("filtered_tokens", remove_stopwords_udf(df1["tokens"]))
    df1 = df1.withColumn("lemmatized_tokens", lemmatize_udf(df1["filtered_tokens"]))

    # Apply the same feature extraction steps
    df1 = df1.withColumn("tweet_length", length(col("text")))
    df1 = df1.withColumn(
        "num_mentions",
        size(split(col("text"), "@")) - 1
    )
    df1 = df1.withColumn(
        "num_hashtags",
        size(split(col("text"), "#")) - 1
    )

    # Transform the lemmatized tokens using the fitted CountVectorizer model
    df1_vectorized = cv_model.transform(df1)

    # Assemble the features using the fitted VectorAssembler model (assuming 'assembler' is defined globally as before for NB)
    # Note: Ensure the 'assembler' used here includes all features used for NB training
    test_features = assembler.transform(df1_vectorized)

    # Make predictions using the trained Naive Bayes model
    test_predict = model.transform(test_features)

    # Get the prediction result
    df_res = test_predict.select('prediction').toPandas()
    pred = df_res.prediction.values[0]

    if pred == 0.0: # Predictions from Naive Bayes are floats
        return "Positive tweet!"
    elif pred == 1.0:
        return "Negative Tweet!"
    return "Neutral tweet"
```

```python
def make_prediction_using_LogReg(user_input):
    # Create a Spark DataFrame from the user input
    df1 = spark.createDataFrame([(1, user_input)], ['Id', 'text'])

    # Apply the same preprocessing steps as the training data
    df1 = df1.withColumn("clean_text", clean_udf(col("text")))
    df1 = df1.withColumn("tokens", tokenize_udf(col("clean_text")))
    df1 = df1.withColumn("filtered_tokens", remove_stopwords_udf(df1["tokens"]))
    df1 = df1.withColumn("lemmatized_tokens", lemmatize_udf(df1["filtered_tokens"]))

    # Apply the same feature extraction steps
    df1 = df1.withColumn("tweet_length", length(col("text")))
    df1 = df1.withColumn(
        "num_mentions",
        size(split(col("text"), "@")) - 1
    )
    df1 = df1.withColumn(
        "num_hashtags",
        size(split(col("text"), "#")) - 1
    )

    # Transform the lemmatized tokens using the fitted CountVectorizer model
    df1_vectorized = cv_model.transform(df1)

    # Assemble the features using the fitted VectorAssembler model for LR (assuming 'assembler' is defined globally as before for LR)
    # We need to redefine the assembler here to match the one used for LR training if it was different from the NB assembler
    lr_assembler = VectorAssembler(
        inputCols=["features", "tweet_length", "num_hashtags", "num_mentions"],
        outputCol="final_features_1" # Use the same output column name as in LR training
    )
    test_features = lr_assembler.transform(df1_vectorized)


    # Make predictions using the trained Logistic Regression model
    lr_predictions = lr_model.transform(test_features)


    # Get the prediction result
    df_res = lr_predictions.select('prediction').toPandas()
    pred = df_res.prediction.values[0]

    if pred == 0.0: # Predictions from Logistic Regression are floats
        return "Positive tweet!"
    elif pred == 1.0:
        return "Negative Tweet!"
    return "Neutral tweet"
```

```python
# Make a prediction using make_prediction_using_LogReg "It was a bad approach"
# YOUR CODE HERE
# Call the prediction function with a sample tweet
result = make_prediction_using_LogReg("It was a bad approach")

# Print the result
print(result)
```
